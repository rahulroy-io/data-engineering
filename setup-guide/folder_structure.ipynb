```python
def generate_names(client, env, domain, entity, source, target, action):
    entity_clean = entity.replace("-", "").replace("_", "").lower()
    return {
        "glue_job": f"gluejob_{env}_{client}_{domain}_{entity_clean}_{source}_to_{target}_{action}",
        "glue_workflow": f"gluewf_{env}_{client}_{domain}_{entity_clean}_{source}_to_{target}",
        "glue_crawler": f"gluecr_{env}_{client}_{domain}_{entity_clean}_{source}",
        "athena_db_ingest": f"{client}_{env}_ingest",
        "athena_db_structured": f"{client}_{env}_structured",
        "athena_db_metadata": f"{client}_{env}_metadata",
        "s3_ingest_path": f"s3://{client}-{env}-ingest/{domain}/{entity_clean}/{{YYYY}}/{{MM}}/{{DD}}/",
        "s3_structured_path": f"s3://{client}-{env}-structured/{domain}/{entity_clean}/{{partition}}/",
        "s3_metadata_runs": f"s3://{client}-{env}-metadata/runs/{domain}/{entity_clean}/run_date={{YYYY-MM-DD}}/",
        "s3_metadata_configs": f"s3://{client}-{env}-metadata/configs/",
        "s3_metadata_schemas": f"s3://{client}-{env}-metadata/schemas/",
        "s3_metadata_lineage": f"s3://{client}-{env}-metadata/lineage/"
    }

# Example usage:
names = generate_names(
    client="myclient",
    env="prod",
    domain="sales",
    entity="orders",
    source="jdbc",
    target="snowflake",
    action="ingest"
)
print(names["glue_job"])  # gluejob_prod_myclient_sales_orders_jdbc_to_snowflake_ingest
```
